```python
#chat prompt
from langchain_core.prompts import ChatPromptTemplate

system_template = "你是一个{role},请用{styele}的风格回答问题{question}"
user_template = "请用简单易懂的方式回答{question}"
#实例化对象
chat_template = ChatPromptTemplate.from_messages(
    [
        ("system", system_template),  
        ("human", user_template)
    ]
)
#填充变量
filled_chat_prompt = chat_template.format_messages(styele="通俗易懂", role="数学教学助手", question="解释勾股定理")
msg = chat.invoke(filled_chat_prompt)
print(msg)
```
