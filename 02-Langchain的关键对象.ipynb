## 输出格式化

```python
from langchain_classic.output_parsers import StructuredOutputParser, ResponseSchema
from langchain_core.prompts import PromptTemplate

# 加载chat
zhipuai_api_key = "4c436c37955741c290d3b97fce80fbb9.R6C2peSTtc9Stc8N"

from langchain_community.chat_models import ChatZhipuAI
from langchain_core.messages import AIMessage, HumanMessage, SystemMessage

import os

os.environ["ZHIPUAI_API_KEY"] = zhipuai_api_key     

chat = ChatZhipuAI(
    model="glm-4",
    temperature=0.5,
)

#定义输出格式 json格式
response_schemas = [
    ResponseSchema(name="name", description="人的姓名", type="string"),
    ResponseSchema(name="age", description="人的年龄", type="integer"),
    ResponseSchema(name="email", description="人的邮箱", type="string")
]   
#创建输出解释器
output_parser = StructuredOutputParser.from_response_schemas(response_schemas)

template_inf = '''
你是一个信息提取助手，请从以下文本中提取出人的姓名、年龄和邮箱，并按照指定的格式输出。
文本：{input_text}
{format_instructions}
'''

prompt_inf = PromptTemplate(
    template=template_inf,
    partial_variables={"format_instructions": output_parser.get_format_instructions()}
)

input_text = "张三，25岁，邮箱是1425@aa.com"
filled_prompt_inf = prompt_inf.format(input_text=input_text)

resp = chat.invoke(filled_prompt_inf)
print(resp)
output_parser.parse(resp.content)
```

```python
from pydantic import BaseModel, Field
from langchain_classic.output_parsers import PydanticOutputParser

class person(BaseModel):
    name: str = Field(description="人的姓名")
    age: int = Field(description="人的年龄")
    email: str = Field(description="人的邮箱")

#创建输出解释器 特定格式输出
pydantic_parser = PydanticOutputParser(pydantic_object=person)

template_inf = '''
你是一个信息提取助手，请从以下文本中提取出人的姓名、年龄和邮箱，并按照person的格式输出。
文本：{input_text}
{format_instructions}
'''

prompt_inf = PromptTemplate(
    template=template_inf,
    partial_variables={"format_instructions": output_parser.get_format_instructions()}
)

input_text = "张三，25岁，邮箱是1425@aa.com"
filled_prompt_inf = prompt_inf.format(input_text=input_text)

msg = chat.invoke(filled_prompt_inf)
print(msg)
pydantic_parser.parse(chat.invoke(filled_prompt_inf).content)
```

## 链式调用

```python
from pydantic import BaseModel, Field
from langchain_classic.output_parsers import PydanticOutputParser
from langchain_core.prompts import PromptTemplate
from langchain_community.chat_models import ChatZhipuAI
from langchain_classic.chains import LLMChain
import os
zhipuai_api_key = "4c436c37955741c290d3b97fce80fbb9.R6C2peSTtc9Stc8N"
## 定义输出格式

class QA(BaseModel):
    answer: str = Field(description="答案")
    confidence: str = Field(description="置信度")
    
## 创建输出解释器
QA_pydantic_parser = PydanticOutputParser(pydantic_object=QA)

templete_qa = '''
你是一名数学老师，请用{style}风格回答以下问题，并以QA的格式返回答案和置信度:
问题:{question}
{format_instructions}
'''

prompt = PromptTemplate(template=templete_qa, 
                        partial_variables={"format_instructions": QA_pydantic_parser.get_format_instructions()})

## 加载模型


os.environ["ZHIPUAI_API_KEY"] = zhipuai_api_key     

chat = ChatZhipuAI(
    model="glm-4",
    temperature=0.5,
)

## 创建链
llm_chain = LLMChain(llm=chat, prompt=prompt, output_parser=QA_pydantic_parser)
#llm_chain = prompt | chat | QA_pydantic_parser #推荐写法
llm_chain = prompt | chat
## 运行链
ai_msg = llm_chain.invoke({
    "style": "通俗易懂",
    "question": "什么是勾股定理？" 
})
ai_msg

print(QA_pydantic_parser.parse(ai_msg.content).answer)
```

## 流式输出  

```python
from pydantic import BaseModel, Field
from langchain_classic.output_parsers import PydanticOutputParser
from langchain_core.prompts import PromptTemplate
from langchain_community.chat_models import ChatZhipuAI
from langchain_classic.chains import LLMChain
import os
zhipuai_api_key = "4c436c37955741c290d3b97fce80fbb9.R6C2peSTtc9Stc8N"
## 定义输出格式

class QA(BaseModel):
    answer: str = Field(description="答案")
    confidence: str = Field(description="置信度")
    
## 创建输出解释器
QA_pydantic_parser = PydanticOutputParser(pydantic_object=QA)

templete_qa = '''
你是一名数学老师，请用{style}风格回答以下问题，并以QA的格式返回答案和置信度:
问题:{question}
{format_instructions}
'''

prompt = PromptTemplate(template=templete_qa, 
                        partial_variables={"format_instructions": QA_pydantic_parser.get_format_instructions()})

## 加载模型


os.environ["ZHIPUAI_API_KEY"] = zhipuai_api_key     

chat = ChatZhipuAI(
    model="glm-4",
    temperature=0.5,
)

## 创建链
llm_chain = prompt | chat 
## 运行链
chunks=[]
for chunk in llm_chain.stream({
    "style": "通俗易懂",
    "question": "什么是勾股定理？" 
}):
    chunks.append(chunk)
    print(chunk.content,end='',flush=True)
```
