## ConversationBufferMemory

```python
from langchain_classic.memory import ConversationBufferMemory
from langchain_classic import ConversationChain

#创建一个记忆系统
memory = ConversationBufferMemory()
#创建会话链
conversation = ConversationChain(llm=chat, memory=memory)
```

```python
conversation.predict(input="你好！我是一个外卖员")
conversation.predict(input="你能否介绍一下自己吗？")
conversation.predict(input="我的职业是？")
```

## ConversationBufferWindowMemory  
  

```python
from langchain_classic.memory import ConversationBufferWindowMemory
window_memory = ConversationBufferWindowMemory(k=1)

conversation_window = ConversationChain(llm=chat, memory=window_memory)
conversation_window.predict(input="你好！我是一个外卖员")
```

## ConversationTokenBufferMemory  
  

```python
from langchain_classic.memory import ConversationTokenBufferMemory
token_memory = ConversationTokenBufferMemory(llm=chat, max_token_limit=100)
token_memory.save_context({"input": "你好！我是一个外卖员"}, {"output": "你好！很高兴认识你！"})
print(token_memory.load_memory_variables({}))
```

```python
print(token_memory.buffer)
```

## ConversationSummaryBufferMemory  
  

```python
from langchain_classic.memory import ConversationSummaryMemory
summary_memory = ConversationSummaryMemory(llm=chat, max_token_limit=100)  
summary_memory.save_context({"input": "你好！我是一个外卖员"}, {"output": "你好！很高兴认识你！"})
print(summary_memory.load_memory_variables({}))
print(summary_memory.buffer)
```

```python
from langchain_classic import ConversationChain
summary_conversation = ConversationChain(llm=chat, memory=summary_memory)
print(summary_conversation.predict(input="我的职业是？"))
```
